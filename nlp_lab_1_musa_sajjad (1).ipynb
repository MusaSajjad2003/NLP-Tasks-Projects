**Task 08 (a): Define regular clean(text) function and an equivalent clean_lambda, testing both on raw string input.**

**Task 08 (b): Define tokenize(text) and remove_short(tokens, n), then chain them on a sample sentence with n=3.**

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence = "I love working with NLP tools in Python daily"
cleaned = clean(sentence)
tokens = tokenize(cleaned)
filtered = remove_short(tokens, 3)
print("Filtered tokens:", filtered)
```

**Task 08 (c): Write word_lengths(tokens) to create a dictionary mapping words to their character length.**

```python
def word_lengths(tokens):
    return {word: len(word) for word in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]
print("Word lengths:", word_lengths(tokens))
```

**Task 08 (d): Combine functions into nlp_pipeline(text, min_len) and test on two different input sentences.**

```python
def nlp_pipeline(text, min_len):
    c_text = clean(text)
    toks = tokenize(c_text)
    filt_toks = remove_short(toks, min_len)
    return word_lengths(filt_toks)

print("Pipe 1:", nlp_pipeline("Text processing with Python is very effective", 3))
print("Pipe 2:", nlp_pipeline("Deep learning models analyze large text datasets", 4))
```

**Task 09 (d): Use Counter to build word frequency from sentence, print top 3, frequency of 'python', total words, and vocab size.**
